```python
import os
from crewai import Agent, Task, Crew

os.environ["OPENAI_API_KEY"] = "" # Replace YOUR_API_KEY with your actual OpenAI API key

def run_project_analysis(problem):

    planner = Agent(
        role="Delivery Planner",
        goal="Create project delivery strategy",
        backstory="A meticulous planner, skilled in structuring complex projects and defining clear, actionable steps."
    )

    analyst = Agent(
        role="Business Analyst",
        goal="Analyze project data",
        backstory="A sharp business analyst with a keen eye for data, able to distill insights and define project requirements."
    )

    risk_manager = Agent(
        role="Risk Manager",
        goal="Identify delivery risks",
        backstory="An experienced risk manager, adept at foreseeing potential issues and developing mitigation strategies."
    )

    task = Task(
        description=problem,
        agent=analyst,
        expected_output="A detailed report analyzing the project data and outlining the requirements for generating a chess game for kids."
    )

    crew = Crew(
        agents=[planner, analyst, risk_manager],
        tasks=[task]
    )

    result = crew.kickoff()

    return result

run_project_analysis(problem="Generate a chess game for kids")
```

```python
import os
from crewai import Agent, Task, Crew

# Set API Key
os.environ["OPENAI_API_KEY"] = ""

####################################################
# PROJECT IDEA
####################################################

project_description = """
Design and implement a secure Login Screen with Single Sign-On (SSO)
for a web application.

Requirements:
- User clicks "Login with SSO"
- Redirect to Identity Provider (Azure AD / Okta)
- OAuth2 authentication
- JWT token validation
- User redirected to dashboard
"""

####################################################
# AGENTS
####################################################

# Business Analyst
ba = Agent(
    role="Business Analyst",
    goal="Analyze product requirements and convert them into user stories",
    backstory="Experienced BA specializing in enterprise authentication systems.",
    verbose=True
)

# Architect
architect = Agent(
    role="Solution Architect",
    goal="Design system architecture for secure SSO login",
    backstory="Enterprise architect specializing in OAuth2, SAML and secure identity management.",
    verbose=True
)

# Frontend Developer
frontend_dev = Agent(
    role="Frontend Developer",
    goal="Develop login screen UI",
    backstory="Expert frontend developer experienced in React and modern UI frameworks.",
    verbose=True
)

# Backend Developer
backend_dev = Agent(
    role="Backend Developer",
    goal="Develop authentication APIs and SSO integration",
    backstory="Backend engineer experienced in OAuth2 authentication and secure API development.",
    verbose=True
)

# QA Engineer
qa = Agent(
    role="QA Engineer",
    goal="Create test strategy and test cases",
    backstory="Automation testing specialist ensuring application quality and security.",
    verbose=True
)

# DevOps
devops = Agent(
    role="DevOps Engineer",
    goal="Deploy application using CI/CD pipeline",
    backstory="DevOps engineer specializing in Docker, Kubernetes and secure deployments.",
    verbose=True
)

####################################################
# TASKS
####################################################

# BA Task
ba_task = Task(
    description=f"""
Analyze the following project request and produce:

1. User stories
2. Functional requirements
3. Acceptance criteria

Project:
{project_description}
""",
    expected_output="Complete requirement document",
    agent=ba
)

# Architecture Task
architecture_task = Task(
    description="""
Design system architecture for SSO login.

Include:
- OAuth2 authentication flow
- Identity provider integration
- token validation
- security architecture
""",
    expected_output="Architecture design document",
    agent=architect
)

# Frontend Task
frontend_task = Task(
    description="""
Design login screen UI.

Include:
- Login with SSO button
- redirect to identity provider
- UI layout
- example React code snippet
""",
    expected_output="UI design and frontend code",
    agent=frontend_dev
)

# Backend Task
backend_task = Task(
    description="""
Develop backend logic for SSO login.

Include:
- OAuth2 callback endpoint
- token validation
- user session creation
- example API code
""",
    expected_output="Backend API design and code example",
    agent=backend_dev
)

# QA Task
qa_task = Task(
    description="""
Create QA testing strategy.

Include:
- functional test cases
- security testing
- SSO authentication tests
- automation strategy
""",
    expected_output="Test plan and test cases",
    agent=qa
)

# DevOps Task
devops_task = Task(
    description="""
Design deployment architecture.

Include:
- Docker configuration
- CI/CD pipeline
- secure environment variables
- cloud deployment
""",
    expected_output="Deployment architecture and pipeline design",
    agent=devops
)

####################################################
# CREW
####################################################

crew = Crew(
    agents=[
        ba,
        architect,
        frontend_dev,
        backend_dev,
        qa,
        devops
    ],
    tasks=[
        ba_task,
        architecture_task,
        frontend_task,
        backend_task,
        qa_task,
        devops_task
    ],
    verbose=True
)

####################################################
# RUN PROJECT
####################################################

result = crew.kickoff()

print("\n\nFINAL SDLC OUTPUT\n")
print(result)
```
