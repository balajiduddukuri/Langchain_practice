```
from google import genai
from google.genai import types
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
client = genai.Client(api_key=userdata.get('GEMINI_API_KEY'))

texts = [
    "What is the meaning of life?",
    "What is the purpose of existence?",
    "How do I bake a cake?",
]

result = client.models.embed_content(
    model="gemini-embedding-001",
    contents=texts,
    config=types.EmbedContentConfig(task_type="SEMANTIC_SIMILARITY")
)

# Create a 3x3 table to show the similarity matrix
df = pd.DataFrame(
    cosine_similarity([e.values for e in result.embeddings]),
    index=texts,
    columns=texts,
)

print(df)
```

```
from google import genai
from google.genai import types

client = genai.Client(api_key=userdata.get('GEMINI_API_KEY'))

# Load your image and set up your prompt
with open('generated_image.png', 'rb') as f:
    image_bytes = f.read()

prompt = """
      Return bounding boxes as a JSON array with labels. Never return masks
      or code fencing. Limit to 25 objects. Include as many objects as you
      can identify on the table.
      If an object is present multiple times, name them according to their
      unique characteristic (colors, size, position, unique characteristics, etc..).
      The format should be as follows: [{"box_2d": [ymin, xmin, ymax, xmax],
      "label": <label for the object>}] normalized to 0-1000. The values in
      box_2d must only be integers
      """

image_response = client.models.generate_content(
  model="gemini-robotics-er-1.5-preview",
  contents=[
    types.Part.from_bytes(
      data=image_bytes,
      mime_type='image/jpeg',
    ),
    prompt
  ],
  config = types.GenerateContentConfig(
      temperature=0.5,
      thinking_config=types.ThinkingConfig(thinking_budget=0)
  )
)

print(image_response.text)
```

## Draw Bounding Boxes and Labels

### Subtask:
Use an image processing library (like PIL/Pillow) to draw the bounding boxes and corresponding labels on the loaded image. The coordinates need to be scaled appropriately as they are normalized to 0-1000.


**Reasoning**:
The next step is to implement the drawing of bounding boxes and labels on the loaded image, as outlined in the subtask instructions. This involves importing necessary modules, iterating through the bounding box data, scaling coordinates, and drawing rectangles and text.



```
from PIL import ImageDraw, ImageFont
from IPython.display import display

# Create a Draw object
draw = ImageDraw.Draw(image)

# Determine image dimensions
img_width, img_height = image.size

# Define font and color
try:
    font = ImageFont.truetype("/usr/share/fonts/truetype/liberation/LiberationSans-Regular.ttf", 20)
except IOError:
    font = ImageFont.load_default()

box_color = "red"
text_color = "red"

# Iterate through each bounding box and draw on the image
for item in bounding_box_data:
    box_2d = item['box_2d']
    label = item['label']

    # Extract and scale coordinates
    ymin, xmin, ymax, xmax = box_2d

    # Coordinates are normalized to 0-1000, scale to image dimensions
    scaled_xmin = int(xmin * (img_width / 1000))
    scaled_ymin = int(ymin * (img_height / 1000))
    scaled_xmax = int(xmax * (img_width / 1000))
    scaled_ymax = int(ymax * (img_height / 1000))

    # Draw rectangle
    draw.rectangle([scaled_xmin, scaled_ymin, scaled_xmax, scaled_ymax], outline=box_color, width=2)

    # Draw label text
    draw.text((scaled_xmin + 5, scaled_ymin + 5), label, fill=text_color, font=font)

# Display the image with bounding boxes
display(image)

print("Bounding boxes and labels drawn successfully on the image.")
```
