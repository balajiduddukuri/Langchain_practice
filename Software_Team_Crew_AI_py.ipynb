```python
# -----------------------------
# Agents
# -----------------------------

ba = Agent(
    role="Business Analyst",
    goal="Gather and document detailed product requirements",
    backstory="Expert in understanding business needs and translating them into technical documentation.",
    verbose=True,
    llm=llm
)

architect = Agent(
    role="Software Architect",
    goal="Design scalable and maintainable system architecture",
    backstory="Experienced architect specializing in cloud-native system design.",
    allow_delegation=True,
    verbose=True,
    llm=llm
)

# Frontend Developers

frontend_dev1 = Agent(
    role="Frontend Developer 1",
    goal="Develop UI components and layout structure",
    backstory="Expert in responsive UI and component design.",
    verbose=True,
    llm=llm
)

frontend_dev2 = Agent(
    role="Frontend Developer 2",
    goal="Implement frontend logic and API integration",
    backstory="Specialist in frontend application logic and API consumption.",
    verbose=True,
    llm=llm
)

# Backend Developers

backend_dev1 = Agent(
    role="Backend Developer 1",
    goal="Develop backend services and APIs",
    backstory="Backend developer skilled in API design and service architecture.",
    verbose=True,
    llm=llm
)

backend_dev2 = Agent(
    role="Backend Developer 2",
    goal="Design and implement database schema",
    backstory="Database expert focused on scalable data architecture.",
    verbose=True,
    llm=llm
)

backend_dev3 = Agent(
    role="Backend Developer 3",
    goal="Implement authentication and security",
    backstory="Security-focused backend engineer.",
    verbose=True,
    llm=llm
)

# QA Testers

qa_tester1 = Agent(
    role="QA Tester 1",
    goal="Test frontend features and user flows",
    backstory="Frontend QA specialist ensuring UI functionality.",
    verbose=True,
    llm=llm
)

qa_tester2 = Agent(
    role="QA Tester 2",
    goal="Test backend APIs and database functionality",
    backstory="Backend QA specialist validating APIs and data operations.",
    verbose=True,
    llm=llm
)

# DevOps

devops = Agent(
    role="DevOps Engineer",
    goal="Deploy application and manage CI/CD pipelines",
    backstory="Infrastructure and deployment automation expert.",
    verbose=True,
    llm=llm
)
```

```python
# -----------------------------
# Tasks
# -----------------------------

ba_task = Task(
    description="Gather detailed requirements for the web application project.",
    expected_output="A comprehensive product requirements document.",
    agent=ba
)

architecture_task = Task(
    description="""
Design the system architecture including:
- technology stack
- frontend architecture
- backend services
- database structure
- API communication
""",
    expected_output="Detailed architecture document.",
    agent=architect,
    context=[ba_task]
)

# Frontend tasks

frontend_task1 = Task(
    description="Design and develop reusable UI components and layouts.",
    expected_output="Frontend UI component structure.",
    agent=frontend_dev1,
    context=[architecture_task]
)

frontend_task2 = Task(
    description="Implement frontend logic, routing, and API integrations.",
    expected_output="Frontend application logic plan.",
    agent=frontend_dev2,
    context=[architecture_task]
)

# Backend tasks

backend_task1 = Task(
    description="Develop backend services and REST APIs.",
    expected_output="Backend service architecture.",
    agent=backend_dev1,
    context=[architecture_task]
)

backend_task2 = Task(
    description="Design database schema and data relationships.",
    expected_output="Database schema documentation.",
    agent=backend_dev2,
    context=[architecture_task]
)

backend_task3 = Task(
    description="Implement authentication and security mechanisms.",
    expected_output="Authentication and security design.",
    agent=backend_dev3,
    context=[architecture_task]
)

# QA tasks

qa_test_ui = Task(
    description="Test frontend UI components and user workflows.",
    expected_output="Frontend QA test report.",
    agent=qa_tester1,
    context=[frontend_task1, frontend_task2]
)

qa_test_backend = Task(
    description="Test backend APIs, authentication, and database interactions.",
    expected_output="Backend QA test report.",
    agent=qa_tester2,
    context=[backend_task1, backend_task2, backend_task3]
)

# DevOps task

devops_task = Task(
    description="""
Prepare CI/CD pipeline and deployment infrastructure.
Automate build and deployment process.
""",
    expected_output="Deployment pipeline configuration.",
    agent=devops,
    context=[qa_test_ui, qa_test_backend]
)
```

```python
# -----------------------------
# Crew
# -----------------------------

crew = Crew(
    agents=[
        ba,
        architect,
        frontend_dev1,
        frontend_dev2,
        backend_dev1,
        backend_dev2,
        backend_dev3,
        qa_tester1,
        qa_tester2,
        devops
    ],
    tasks=[
        ba_task,
        architecture_task,
        frontend_task1,
        frontend_task2,
        backend_task1,
        backend_task2,
        backend_task3,
        qa_test_ui,
        qa_test_backend,
        devops_task
    ],
    process=Process.sequential,
    verbose=True
)
```

```python
# -----------------------------
# Run Crew
# -----------------------------

if __name__ == "__main__":
    result = crew.kickoff()

    print("\n\nFINAL RESULT\n")
    print(result)
```
