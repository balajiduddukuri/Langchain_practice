### Case Study: Analyzing Website Traffic Logs

Imagine you are a data analyst or site reliability engineer (SRE) responsible for monitoring a website's performance and user activity. You regularly receive large log files (e.g., Apache or Nginx access logs) that contain detailed information about every request made to your server.

This is a perfect scenario to apply concepts from Week 4:

1.  **File I/O (Reading Log Files):**
    - You'd use `open()` and `with` statements to read these large log files line by line, as direct `input()` is not feasible for such data sources.
    - Example: `with open('access_log.txt', 'r') as log_file:`

2.  **Error Handling (Handling Malformed Log Entries):**
    - Log files can sometimes contain corrupted or malformed entries due to various reasons. When parsing each line, converting strings to integers (e.g., status codes) or floats (e.g., response times) could raise `ValueError`.
    - You would implement `try-except` blocks to gracefully handle these errors, skipping problematic lines or logging them for later inspection, instead of crashing your analysis script.

3.  **Functional Programming (`map`, `filter`, `reduce`):**
    - **`map()`:** After reading each line, you might use `map()` with a `lambda` function or a custom parsing function to transform raw log strings into structured data (e.g., extract IP address, timestamp, HTTP method, URL, status code, response time for each request).
    - **`filter()`:** You could use `filter()` to select specific types of requests for analysis, such as:
        - All requests with a 404 (Not Found) status code to identify broken links.
        - All requests from a specific IP range to monitor bot activity.
        - Requests for images (`.jpg`, `.png`) vs. HTML pages.
    - **`reduce()`:** To aggregate data, you might use `reduce()` to calculate the total data transferred, sum up response times, or count unique visitors over a period.

4.  **NumPy (Numerical Analysis of Metrics):**
    - Once you've extracted numerical data (like response times, data transfer sizes, or even counts of events per minute/hour), you'd load this into NumPy arrays.
    - **Fast Operations:** NumPy's vectorized operations would allow you to quickly calculate statistics like average response time (`np.mean()`), maximum data transfer (`np.max()`), standard deviation of response times (`np.std()`), or filter out outliers without writing explicit Python loops, which would be much slower for large datasets.
    - **Data Structures:** You could create 2D NumPy arrays to store time-series data of various metrics (e.g., `[timestamp, successful_requests, 4xx_errors, 5xx_errors]`) for further analysis or visualization.

This integrated approach allows you to efficiently process, clean, analyze, and extract insights from vast amounts of website log data, which is critical for maintaining a healthy and performant online service.

Correlation measures the strength and direction of a relationship between two numerical values.

Correlation is measured by **Correlation Cofficient**: -1 to +1

- +1 -> Perfect Correlation Coefficient
- -1 -> Perfect Negative Correlation
- 0 -> No linear correlation

Positive Correlation (r > 0)
- As one variable increases the other variable also increases
- hours studied (increases) -> marks (increases)

Negative Correlation (r<0)
- As one variable increases -> the other decreases
- eg., Speed (increases) -> Time taken (decreases)

Zero Correlation (r ~ 0)
- No linear relationship
- Shoe size and income

Correlation != Causation

````python
from langchain_community.vectorstores import Chroma
from langchain_openai import ChatOpenAI, OpenAIEmbeddings

# Initialize Embeddings:
embeddings = OpenAIEmbeddings()

# Create/ Load Vector Databases:
vectorstore = Chroma(
    collection_name = "long_term_memory",
    embedding_function = embeddings,
    persist_directory = "./memory_db"
)
````

### Vector DB:
Text -> Converted to numbers (vectors)

A vector database is a database that is designed to Store the Vectors (embeddings) and quickly find the similar vectors.

- Find the content similar to the meaning.

Compare Traditional DB vs Vector DB

1. Data Type
- Traditional DB: Text, Numbers -> "AI Course"
- Vector DB: Vectors(embeddings) -> [0.23, 0.35, ...]

2. Search Type:
- Traditional DB: Exact Match
- Vector DB: Seamantic (meaning-based)

3. Intelligence:
- Traditional DB: No Meaning / Synonyms -> Fails
- Vector DB: Understands Meaning / Synonyms -> Works

- Traditional DB -> Banking Systems/ User records/ Transctions/ Inventory
- Vector DB -> Semantic Search/ Chatbots/ Recommendation Systems/ RAG (LLM apps)

- Exact Match -> Traditional DB
- Match the meaning -> Vector DB

A Vectordatabse stores embeddings and retrieves data based on meaning rather than exact matches.

### Store Long Term Memories

````python
def store_long_term_memories(text, user_id):
  vectorstore.add_texts(
      texts = [text],
      metadatas = [{"user_id": user_id}]
  )
````

## Retrieve Relevant Long-Term Memory

````python
def retrieve_long_term_memory(query, user_id, k = 1):
  docs = vectorstore.similarity_search(
      query,
      k = k,
      filter = {"user_id": user_id}
  )
  return "\n".join([doc.page_content for doc in docs])
````

1 = Exactly same direction  -> Similar meaning
0 = No relation (Orthogonal) -> Unrelated
-1 = Opposite meaning -> Opposite Meaning

````python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model = "gpt-4o-mini",
    temperature = 0
)
````

````python
from langchain_core.tools import tool

@tool
def add_numbers(a: int, b: int) -> int:
  """Adds two Numbers"""
  return a + b

@tool
def multiply_numbers(a: int, b: int) -> int:
  """Multiply 2 nos."""
  return a * b
````

````python
tools = [add_numbers, multiply_numbers]
````

````python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt = ChatPromptTemplate(
    [(
    "system",
    """Answer the following questions as best you can.
You have access to the following tools.

Use the following format internally:

Question: the input question you must answer
Thought: you should always think about what to do
Action: the action to take
Action Input: the input to the action
Observation: the result of the action
(repeat as needed)
Thought: I now know the final answer
Final Answer: the final answer to the original input question
"""
  ),
  MessagesPlaceholder(variable_name = "messages"),
])
````

````python
from langchain_core.tools import tool
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langgraph.prebuilt import create_react_agent

import wikipedia

@tool
def wikipedia_search(aquery: str) -> str:
    """Search Wikipedia and return a short summary"""
    try:
        return wikipedia.summary(aquery, sentences=2)
    except Exception as e:
        return f"Wikipedia error: {str(e)}"

@tool
def calculator(expression: str) -> str:
    """Evaluate a mathematical expression"""
    try:
        return str(eval(expression))
    except Exception as e:
        return f"Calculation error: {str(e)}"

tools = [wikipedia_search, calculator]


prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """
        You are an intelligent research scientist

        RULES:
        1. Use wikipedia once to get factual information
        2. Use calculator if numerical calculation is required
        3. After using tools, produce final answer
        4. Do not repeat tool usage after final answer
        """
    ),
    MessagesPlaceholder(variable_name="messages"),
])


agent = create_react_agent(
    model=llm,
    tools=tools,
    prompt=prompt
)

result = agent.invoke({
    "messages": [
        ("user", "What is the population of India according to Wikipedia, what is 20% of it?")
    ]
})

print(result["messages"][-1].content)
````
